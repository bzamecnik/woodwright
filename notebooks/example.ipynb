```python
import build123d as bd
from jupyter_cadquery import set_defaults

import woodwright as ww
```

```python
set_defaults(transparent=True, default_opacity=0.4)
```

```python
oak = ww.Stock.board(20, "oak")
root_location = bd.Location(bd.Plane((0, 20, 0), x_dir=(0, 0, 1), z_dir=(0, 1, 0)))
```

## Mortise-Tenon Joint

```python
asm = ww.Assembly("mortise_tenon", root_location=root_location)
stile_length = 600
rail_length = 300
board_width = 60
stile = asm.add(ww.Board("stile", oak, length=stile_length, width=board_width))
rail = asm.add(ww.Board("rail", oak, length=rail_length, width=board_width))
asm.join(ww.MortiseTenon(tenon=rail.left_end, mortise=stile.back_edge, tenon_length=board_width))

model = asm.resolve()
model.compound()
```

## Mortise-Tenon Frame

```python
asm = ww.Assembly("frame", root_location=root_location)
shoulder = 10
stile_left = asm.add(ww.Board("stile_left", oak, length=stile_length, width=board_width))
stile_right = asm.add(ww.Board("stile_right", oak, length=stile_length, width=board_width))
rail_top = asm.add(ww.Board("rail_top", oak, length=rail_length, width=board_width))
rail_bottom = asm.add(ww.Board("rail_bottom", oak, length=rail_length, width=board_width))
asm.join(
    ww.MortiseTenon(
        tenon=rail_top.left_end,
        mortise=stile_left.back_edge,
        position=ww.Offset(0),
        shoulder=shoulder,
    )
)
asm.join(
    ww.MortiseTenon(
        tenon=rail_top.right_end,
        mortise=stile_right.front_edge,
        position=ww.Offset(stile_right.length - rail_top.width),
        shoulder=shoulder,
    )
)
asm.join(
    ww.MortiseTenon(
        tenon=rail_bottom.left_end,
        mortise=stile_left.back_edge,
        position=ww.Offset(stile_left.length - rail_bottom.width),
        shoulder=shoulder,
    )
)
asm.join(
    ww.MortiseTenon(
        tenon=rail_bottom.right_end,
        mortise=stile_right.front_edge,
        position=ww.Offset(0),
        shoulder=shoulder,
    )
)

model = asm.resolve()
model.compound()
```
